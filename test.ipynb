```python
import matplotlib.pyplot as plt

# 데이터 설정
x = [1, 2, 3, 4, 5]
y = [i**2 for i in x]

# 그래프 그리기
plt.figure(figsize=(6, 4))
plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.title('Colab 예제 시각화')
plt.xlabel('X 값')
plt.ylabel('Y 값')
plt.grid(True)
plt.show()
```
